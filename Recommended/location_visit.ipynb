```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\workspace\Ko-Swipe-ML\data\preprocessing\관광지 추천시스템 Trainset_B.csv")
df1 = pd.read_csv(r"C:\workspace\Ko-Swipe-ML\data\preprocessing\관광지 추천시스템 Testset_B.csv")
```

```python
df = df[["TRAVEL_ID", "VISIT_AREA_NM"]]
df1 = df1[["TRAVEL_ID", "VISIT_AREA_NM"]]
```

```python
df = pd.concat([df, df1]).reset_index(drop=True)
```

```python
df.to_csv(r"C:\workspace\Ko-Swipe-ML\data\preprocessing\followup_places.csv", index=False)
```

```python
haeundae_visitors = df[df['VISIT_AREA_NM'].str.contains('광안리해수욕장')]['TRAVEL_ID'].unique()
```

```python
other_visits_by_haeundae_visitors = df[df['TRAVEL_ID'].isin(haeundae_visitors)]
other_visits_by_haeundae_visitors
```

```python
other_visits_excluding_haeundae = other_visits_by_haeundae_visitors[
    ~other_visits_by_haeundae_visitors['VISIT_AREA_NM'].str.contains('광안리해수욕장')
]
other_visits_excluding_haeundae
```

```python
popular_places_after_haeundae = other_visits_excluding_haeundae['VISIT_AREA_NM'].value_counts().reset_index()
popular_places_after_haeundae.columns = ['VISIT_AREA_NM', 'visit_count']
```
